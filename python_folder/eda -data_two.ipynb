# Comprehensive data cleaning pipeline
class DataCleaningPipeline:
    def __init__(self, dataframe):
        self.df = dataframe
        self.cleaning_log = []
    
    def handle_missing_values(self):
        # Numeric columns: mean imputation
        numeric_cols = self.df.select_dtypes(include=[np.number]).columns
        for col in numeric_cols:
            if self.df[col].isnull().sum() > 0:
                mean_val = self.df[col].mean()
                self.df[col].fillna(mean_val, inplace=True)
                self.cleaning_log.append(f"Imputed {col} with mean: {mean_val:.2f}")
        
        # Categorical columns: mode imputation or 'Unknown'
        categorical_cols = self.df.select_dtypes(include=['object']).columns
        for col in categorical_cols:
            if self.df[col].isnull().sum() > 0:
                mode_val = self.df[col].mode()[0] if not self.df[col].mode().empty else 'Unknown'
                self.df[col].fillna(mode_val, inplace=True)
                self.cleaning_log.append(f"Imputed {col} with mode: {mode_val}")
    
    def remove_duplicates(self):
        initial_count = len(self.df)
        self.df.drop_duplicates(inplace=True)
        final_count = len(self.df)
        removed = initial_count - final_count
        self.cleaning_log.append(f"Removed {removed} duplicate records")
    
    def standardize_formats(self):
        # Standardize column names
        self.df.columns = self.df.columns.str.lower().str.replace(' ', '_')
        
        # Standardize text data
        text_columns = self.df.select_dtypes(include=['object']).columns
        for col in text_columns:
            self.df[col] = self.df[col].str.strip().str.title()
        
        self.cleaning_log.append("Standardized column names and text formats")
```

class DataIntegrationEngine:
    def __init__(self):
        self.integration_log = []
        self.merged_datasets = {}
    
    def merge_weekly_data(self, weekly_files):
        """Merge multiple weekly data files into longitudinal dataset"""
        merged_df = pd.DataFrame()
        
        for week_num, file_path in enumerate(weekly_files, 1):
            try:
                week_data = pd.read_csv(file_path)
                week_data['week'] = week_num
                merged_df = pd.concat([merged_df, week_data], ignore_index=True)
                self.integration_log.append(f"✓ Integrated Week {week_num} data")
            except Exception as e:
                self.integration_log.append(f"✗ Error integrating Week {week_num}: {e}")
        
        return merged_df
    
    def merge_demographic_datasets(self, demo_datasets):
        """Integrate multiple demographic datasets"""
        master_demo = pd.DataFrame()
        
        for dataset_name, dataset in demo_datasets.items():
            # Align schemas before merging
            aligned_dataset = self.align_schema(dataset, master_demo)
            master_demo = pd.concat([master_demo, aligned_dataset], ignore_index=True)
            self.integration_log.append(f"✓ Integrated {dataset_name}")
        
        return master_demo
    
    def align_schema(self, source_df, target_df):
        """Align schemas between datasets for consistent merging"""
        if target_df.empty:
            return source_df
        
        # Add missing columns with default values
        for col in target_df.columns:
            if col not in source_df.columns:
                source_df[col] = None
        
        # Reorder columns to match target schema
        source_df = source_df[target_df.columns]
        return source_df
```
