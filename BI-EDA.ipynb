```python
import numpy as np


columns_to_clean = [
    'Satisfied with Course',
    'Satisfied with Teaching',
    'Satisfied with Feedback',
    'Student to Staff Ratio',
    'Spend per Student'
]

# ---------- Clean institution_df ----------
for col in columns_to_clean:
    institution_df[col] = institution_df[col].replace('-', np.nan)
    institution_df[col] = pd.to_numeric(institution_df[col], errors='coerce')
    institution_df[col] = institution_df[col].fillna(institution_df[col].median())

# ---------- Clean subject_df ----------
for col in columns_to_clean:
    subject_df[col] = subject_df[col].replace('-', np.nan)
    subject_df[col] = pd.to_numeric(subject_df[col], errors='coerce')
    subject_df[col] = subject_df[col].fillna(subject_df[col].median())
```

```python
print("Institution nulls:\n", institution_df[columns_to_clean].isnull().sum())
print("Subject nulls:\n", subject_df[columns_to_clean].isnull().sum())

# Preview cleaned data
print("\nInstitution sample:")
print(institution_df[['Institution', 'Year'] + columns_to_clean].head())

print("\nSubject sample:")
print(subject_df[['Institution', 'Year', 'Subject'] + columns_to_clean].head())
```

```python
# For institution_df
institution_df['Average Entry Tariff'] = institution_df['Average Entry Tariff'].replace('-', np.nan)
institution_df['Average Entry Tariff'] = pd.to_numeric(institution_df['Average Entry Tariff'], errors='coerce')
institution_df['Average Entry Tariff'] = institution_df['Average Entry Tariff'].fillna(institution_df['Average Entry Tariff'].median())

# For subject_df
subject_df['Average Entry Tariff'] = subject_df['Average Entry Tariff'].replace('-', np.nan)
subject_df['Average Entry Tariff'] = pd.to_numeric(subject_df['Average Entry Tariff'], errors='coerce')
subject_df['Average Entry Tariff'] = subject_df['Average Entry Tariff'].fillna(subject_df['Average Entry Tariff'].median())
```

```python
print("Institution missing tariff:", institution_df['Average Entry Tariff'].isnull().sum())
print("Subject missing tariff:", subject_df['Average Entry Tariff'].isnull().sum())

print("Tariff summary (institution):")
print(institution_df['Average Entry Tariff'].describe())
```

```python
# Clean and fill Value Added Score in institution_df
institution_df['Value Added Score'] = institution_df['Value Added Score'].replace('-', np.nan)
institution_df['Value Added Score'] = pd.to_numeric(institution_df['Value Added Score'], errors='coerce')
institution_df['Value Added Score'] = institution_df.groupby('Institution')['Value Added Score']\
    .transform(lambda x: x.fillna(x.median()))

# Clean and fill in subject_df
subject_df['Value Added Score'] = subject_df['Value Added Score'].replace('-', np.nan)
subject_df['Value Added Score'] = pd.to_numeric(subject_df['Value Added Score'], errors='coerce')
subject_df['Value Added Score'] = subject_df.groupby('Institution')['Value Added Score']\
    .transform(lambda x: x.fillna(x.median()))
```

```python
print("Institution nulls (VAS):", institution_df['Value Added Score'].isnull().sum())
print("Subject nulls (VAS):", subject_df['Value Added Score'].isnull().sum())

print(subject_df[['Institution', 'Year', 'Value Added Score']].sort_values(by='Value Added Score').head(10))
```

```python
# Institution-level
institution_df['Career after 15 months'] = institution_df['Career after 15 months'].replace('-', np.nan)
institution_df['Career after 15 months'] = pd.to_numeric(institution_df['Career after 15 months'], errors='coerce')
institution_df['Career after 15 months'] = institution_df.groupby('Institution')['Career after 15 months']\
    .transform(lambda x: x.fillna(x.median()))

# Subject-level
subject_df['Career after 15 months'] = subject_df['Career after 15 months'].replace('-', np.nan)
subject_df['Career after 15 months'] = pd.to_numeric(subject_df['Career after 15 months'], errors='coerce')
subject_df['Career after 15 months'] = subject_df.groupby('Institution')['Career after 15 months']\
    .transform(lambda x: x.fillna(x.median()))
```

```python
print("Remaining nulls:", subject_df['Career after 15 months'].isnull().sum())
print(subject_df[['Institution', 'Year', 'Career after 15 months']].sort_values(by='Career after 15 months').head(10))
```

```python
# Institution-level
institution_df['Continuation'] = institution_df['Continuation'].replace('-', np.nan)
institution_df['Continuation'] = pd.to_numeric(institution_df['Continuation'], errors='coerce')
institution_df['Continuation'] = institution_df.groupby('Institution')['Continuation']\
    .transform(lambda x: x.fillna(x.median()))

# Subject-level
subject_df['Continuation'] = subject_df['Continuation'].replace('-', np.nan)
subject_df['Continuation'] = pd.to_numeric(subject_df['Continuation'], errors='coerce')
subject_df['Continuation'] = subject_df.groupby('Institution')['Continuation']\
    .transform(lambda x: x.fillna(x.median()))
```

```python
print("Nulls left (institution):", institution_df['Continuation'].isnull().sum())
print("Nulls left (subject):", subject_df['Continuation'].isnull().sum())

# Preview sample values
print(institution_df[['Institution', 'Year', 'Continuation']].sort_values(by='Continuation').head())
```

```python
#How has UWE Bristol performed over the period?

uwe_df = institution_df[institution_df['Institution'] == 'UWE Bristol'].sort_values(by='Year')

# Line chart of Guardian Score
plt.figure(figsize=(10, 5))
sns.lineplot(data=uwe_df, x='Year', y='Guardian Score', marker='o')
plt.title("UWE Bristol Guardian Score (2011–2022)")
plt.ylim(0, 100)
plt.show()

# Line chart of Rank
plt.figure(figsize=(10, 5))
sns.lineplot(data=uwe_df, x='Year', y='Rank', marker='o')
plt.gca().invert_yaxis()  # lower rank is better
plt.title("UWE Bristol Guardian Rank (2011–2022)")
plt.ylabel("Rank (Lower is Better)")
plt.show()
```

```python
#How have UWE’s subjects fared over the period?

uwe_subjects = subject_df[subject_df['Institution'] == 'UWE Bristol']

# Average score per subject over time
subject_scores = uwe_subjects.groupby(['Subject', 'Year'])['Guardian Score'].mean().reset_index()

# Optional: Pivot for heatmap
heatmap_df = subject_scores.pivot(index='Subject', columns='Year', values='Guardian Score')

plt.figure(figsize=(14, 10))
sns.heatmap(heatmap_df, cmap="YlGnBu", annot=True, fmt=".1f")
plt.title("UWE Bristol Subject Guardian Scores (2011–2022)")
plt.show()
```

```python
#What’s next for UWE Bristol in the Guardian rankings?

# Linear trendline (score prediction)
import numpy as np
from sklearn.linear_model import LinearRegression

X = uwe_df[['Year']]
y = uwe_df['Guardian Score']

model = LinearRegression()
model.fit(X, y)

future_year = 2023
predicted_score = model.predict([[future_year]])
print(f"📈 Predicted Guardian Score for {future_year}: {predicted_score[0]:.2f}")
```

```python
# How can UWE Bristol perform better in future?

# Compare UWE to top 5 universities in latest year
top5 = institution_df[institution_df['Year'] == 2022].nlargest(5, 'Guardian Score')
uwe_latest = uwe_df[uwe_df['Year'] == 2022]

compare_df = pd.concat([uwe_latest, top5])
metrics = ['Satisfied with Course', 'Satisfied with Teaching', 'Satisfied with Feedback', 
           'Spend per Student', 'Student to Staff Ratio', 'Continuation']

compare_df = compare_df[['Institution'] + metrics].set_index('Institution')
compare_df.T.plot(kind='bar', figsize=(12, 6), title="UWE vs Top 5 Institutions – Key Metrics (2022)")
plt.ylabel("Score / Ratio")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
